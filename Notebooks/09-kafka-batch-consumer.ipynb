```python
BOOTSTRAP_SERVER = dbutils.secrets.get(scope="kafka-secrets",key="bootstrap-server")
JAAS_MODULE =  "kafkashaded.org.apache.kafka.common.security.plain.PlainLoginModule"
CLUSTER_API_KEY = dbutils.secrets.get(scope="kafka-secrets",key="cluster-api-key")
CLUSTER_API_SECRET = dbutils.secrets.get(scope="kafka-secrets",key="cluster-api-secret")
```

```python
df = (
    spark.read
        .format("kafka")
        .option("kafka.bootstrap.servers", BOOTSTRAP_SERVER)
        .option("kafka.security.protocol", "SASL_SSL")
        .option("kafka.sasl.mechanism", "PLAIN")
        .option(
            "kafka.sasl.jaas.config",
            f"{JAAS_MODULE} required "
            f"username='{CLUSTER_API_KEY}' "
            f"password='{CLUSTER_API_SECRET}';"
        )
        .option("subscribe", "invoices")
        .load()
)
```
